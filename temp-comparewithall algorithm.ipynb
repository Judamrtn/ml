importing libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

loading the dataset and spliting into x and y

```python
df = pd.read_csv('Temperature Dataset.csv')
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

spliting into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

Linear regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
pred_lr = lr.predict(X_test)
```

Polynomial

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=4)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
pred_poly = poly_model.predict(X_test_poly)
```

SVR

```python
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

sc_X = StandardScaler()
sc_y = StandardScaler()

X_train_sc = sc_X.fit_transform(X_train)
y_train_sc = sc_y.fit_transform(y_train.reshape(-1,1)).ravel()

svr = SVR(kernel='rbf')
svr.fit(X_train_sc, y_train_sc)

pred_svr = sc_y.inverse_transform(svr.predict(sc_X.transform(X_test)).reshape(-1,1)).ravel()
```

Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(random_state=0)
dt.fit(X_train, y_train)
pred_dt = dt.predict(X_test)
```

Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100, random_state=0)
rf.fit(X_train, y_train)
pred_rf = rf.predict(X_test)
```

Compare all models

```python
print('Linear:', r2_score(y_test, pred_lr))
print('Polynomial:', r2_score(y_test, pred_poly))
print('SVR:', r2_score(y_test, pred_svr))
print('Decision Tree:', r2_score(y_test, pred_dt))
print('Random Forest:', r2_score(y_test, pred_rf))
```
